```python
import re
import requests
from pathlib import Path

API_KEY = "sk-PnmrCHSKUlL46iDPqhqY2DKI43ixLSzQvHgJKhdwyrigx1yx"
BASE_URL = "https://apirouter.ai/v1"

def draw_and_save(prompt: str, output_path: str = "output.png") -> str:
    url = f"{BASE_URL}/chat/completions"
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json",
    }
    payload = {
        "model": "gpt-image-2",
        "messages": [{"role": "user", "content": prompt}],
        "stream": False
    }

    resp = requests.post(url, headers=headers, json=payload, timeout=180)
    resp.raise_for_status()
    data = resp.json()

    content = data["choices"][0]["message"]["content"]
    # 匹配 markdown 图片链接: ![image](https://...)
    match = re.search(r'!\[[^\]]*\]\((https?://[^\s)]+)\)', content)
    if not match:
        raise ValueError(f"未在返回内容中找到图片链接，原始内容: {content}")

    image_url = match.group(1)

    img_resp = requests.get(image_url, timeout=180)
    img_resp.raise_for_status()

    path = Path(output_path)
    path.write_bytes(img_resp.content)
    return str(path.resolve())

if __name__ == "__main__":
    prompt = "生成一张图：一只戴宇航员头盔的橘猫，电影感，1024x1024"
    saved = draw_and_save(prompt, "images/cat_astronaut.png")
    print("图片已保存:", saved)
```

```python
import re
import requests
from pathlib import Path
from openai import OpenAI

client = OpenAI(
    base_url="https://apirouter.ai/v1",  # 建议带 /v1
    api_key="sk-PnmrCHSKUlL46iDPqhqY2DKI43ixLSzQvHgJKhdwyrigx1yx"
)

response = client.chat.completions.create(
    model="gpt-image-2",
    messages=[{
        "role": "user",
        "content": "生成一张图：一只戴宇航员头盔的橘猫，电影感，1024x1024"
    }]
)

content = response.choices[0].message.content or ""
print("模型返回内容：", content)

# 从 markdown 图片中提取 URL：![image](https://xxx.png)
match = re.search(r'!\[[^\]]*\]\((https?://[^\s)]+)\)', content)
if not match:
    raise ValueError(f"未找到图片链接，返回内容: {content}")

image_url = match.group(1)
img_resp = requests.get(image_url, timeout=180)
img_resp.raise_for_status()

output_path = Path("images/gpt_image_2_output.png")
output_path.write_bytes(img_resp.content)
print("图片已保存：", output_path.resolve())
```
